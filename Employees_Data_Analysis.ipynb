```python
# employee who joined the company earliest
earliest_emp = df.loc[df["Joining Date"].idxmin()]
print("Earliest joined employee")
print(earliest_emp)
```

```python
#finding employees earning above dept avg salary
dept_avg_sal = df.groupby("Department")["Salary "].transform("mean")
print(dept_avg_sal)

#filtering employees
above_avg_emp = df[df["Salary "] > dept_avg_sal]
print(above_avg_emp)
```

```python
# creating summary of dataframe with dept wise - min,max,avg-salary
#dept_summary = df.groupby("department")["salary"].agg(
   # min_salary="min",
  #  max_salary="max",
 #   avg_salary="mean"
#).reset_index()

#print("Department-wise Salary Summary:")
#print(dept_summary)

dept_summary = df.groupby("Department")["Salary "].agg(
    min_salary = "min",
    max_salary = 'max',
    avg_salary = "mean"
).reset_index()

print("Department-wise Salary Summary")
print(dept_summary)
```

```python
# finding the department highest avg bonus.
#dept_avg_bonus = df.groupby("department")["bonus"].mean().reset_index()
#dept_highest_bonus = dept_avg_bonus.loc[dept_avg_bonus["bonus"].idxmax()]

#print("Department with highest average bonus:")
#print(dept_highest_bonus)

dept_avg_bonus = df.groupby("Department")["Bonus"].mean().reset_index()

dept_highest_bonus = dept_avg_bonus.loc[dept_avg_bonus["Bonus"].idxmax()]

print("Department with highest average bonus:")
print(dept_highest_bonus)
```
